```python
from rdkit.Chem import AllChem, Draw

SMILES = r"C(CC(=O)N)[C@@H](C(=O)O)N"

mol = AllChem.MolFromSmiles(SMILES)  # Your SMILES here
mol = AllChem.AddHs(mol)
canonical_order = AllChem.CanonicalRankAtoms(mol)

# Assign the rank as a note to each atom
for atom, rank in zip(mol.GetAtoms(), canonical_order):
    atom.SetProp("atomNote", str(rank + 1))

# Compute 2D coordinates for nice visualization
AllChem.Compute2DCoords(mol)

# Draw the molecule
Draw.MolToImage(mol, size=(800, 600))
```
